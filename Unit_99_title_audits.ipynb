```python
import pandas as pd

folder_99_lists = "/content/drive/MyDrive/Data/Unit_99_Title_Claim/Historical 99 Data"

latest_99_file = pd.ExcelFile(f"{folder_99_lists}/UPTE_99_as_of_9_2_2026.xlsx")
latest_99_file.sheet_names
```

```python
latest_99_df = latest_99_file.parse('Report Page')
latest_99_df.shape
```

```python
latest_99_df.columns
```

```python
second_latest_99_file = pd.ExcelFile(f"{folder_99_lists}/UPTE_99_as_of_6_3_2026.xlsx")
second_latest_99_file.sheet_names
```

```python
second_latest_99_df = second_latest_99_file.parse('Sheet1')
second_latest_99_df.shape
```

```python
latest_job_titles_set = set(latest_99_df['Job Code Description'].unique())
```

```python
len(latest_job_titles_set)
```

```python
second_latest_Job_titles_set = set(second_latest_99_df['Job Code Description'].unique())
```

```python
len(second_latest_Job_titles_set)
```

```python
titles_in_june_not_sept_list = second_latest_Job_titles_set - latest_job_titles_set
```

```python
titles_in_sept_not_june_list = latest_job_titles_set - second_latest_Job_titles_set
```

```python
titles_in_june_not_sept_df = second_latest_99_df[second_latest_99_df['Job Code Description'].isin(titles_in_june_not_sept_list)]
titles_in_june_not_sept_df.shape
```

```python
titles_in_june_not_sept_summary = titles_in_june_not_sept_df.pivot_table(values='Employee Lived Name', \
                                      index=['Job Code Description', 'Job Code'], columns=['Location/Business Unit'], \
                                      aggfunc='count', margins=True, margins_name='Total')
titles_in_june_not_sept_summary.fillna("-", inplace=True)
titles_in_june_not_sept_summary
```

```python
titles_in_june_not_sept_summary.to_excel("titles_in_june_not_sept_summary.xlsx")
#
```

```python
sept_99_headcnt_location = latest_99_df.pivot_table(values='Employee Lived Name', \
            index='Job Code Description', columns=['Location/Business Unit'], \
             aggfunc='count').fillna("-")
sept_99_headcnt_location
```

```python
june_99_headcnt_location = second_latest_99_df.pivot_table(values='Employee Lived Name', index='Job Code Description', columns=['Location/Business Unit'], \
                    aggfunc='count').fillna("-")
```

```python
with pd.ExcelWriter("99_headcnt_location.xlsx") as writer:
  sept_99_headcnt_location.to_excel(writer, sheet_name="sept_99_headcnt_location")
  june_99_headcnt_location.to_excel(writer, sheet_name="june_99_headcnt_location")
```
