```python
Healthplan_data_path = "/content/drive/MyDrive/Research/Bargaining Data 2024/Agitational Resources/UPTE RFI 4.12.23. Response to Item No. 20_Medical Benefit Plans(revised) sorted.xlsx"
```

```python
healthplan_file = pd.ExcelFile(Healthplan_data_path)
tab_names = healthplan_file.sheet_names
tab_names
```

```python
# Define the data types for specific columns
dtype_mapping = {
    'Employee ID': str,
    'Union Code': str,
    'Job Code': str
}

# Parse the excel sheet specifying the column data types
healthplan_df = healthplan_file.parse(tab_names[0], dtype=dtype_mapping)
```

```python
healthplan_df.shape
```

```python
healthplan_df['Employee ID'].nunique()
```

```python
healthplan_w_location_df = healthplan_df.merge(juris_2023_df[['Location', 'EE ID']], left_on='Employee ID', right_on='EE ID', how='left').drop_duplicates('Employee ID')
healthplan_w_location_df.shape
```

```python
healthplan_w_location_df['Location'].unique()
```

```python
healthplan_w_location_df['Business Unit'] = healthplan_w_location_df['Location'].map(location_dict)
healthplan_w_location_df.head()
```

```python
healthplan_w_location_df.drop(columns=['EE ID'], inplace=True)
```
