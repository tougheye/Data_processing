This script will unzip folders in Google Drive

```python
import os, shutil

master_folder = "/content/drive/MyDrive/Research/UC Fund Balance/"
latest_data_folder = "July 2026 Common Chart of Accounts (CCOA) input submission files"

dir_in_latest_data_folder = [x for x in os.listdir(f"{master_folder}/{latest_data_folder}") if x.endswith((".zip", ".tar"))]
print(dir_in_latest_data_folder)
```

```python
for folder in dir_in_latest_data_folder:
  dest_path = f"{master_folder}July_2026_balance/{folder.split('.')[0]}"
  shutil.unpack_archive(f"{master_folder}/{latest_data_folder}/{folder}", dest_path)
```
