This script takes in the broadstripe export of doctor surveys and maps the responses based on the specialties

```python
import pandas as pd

# read the csv file exported from Broadstripe
file_path = "/content/Survey Responses 11-11 (3).csv"
response_df = pd.read_csv(file_path)
response_df.shape
```

```python
# see which specialties are leading the chart
response_df['Specialty'].value_counts()
```

```python
# number of specialties
response_df['Specialty'].nunique()      #30
```

```python
# get the analysis survey responses in Survey Question 2 field in a list
survey_cats = ['Compensation','Leaves (parental, medical, sick, etc.)',
               'Protected time','Work/life balance','Staffing', 'Retirement']
```

```python
response_df.columns
```

```python
# Create a copy of the repsonse df to work with
cols_needed = ['Broadstripes ID', 'First Name', 'Last Name',\
               'Employer', 'Department', 'Classification','Survey Question 1',\
              'Survey Question 2','Specialty', 'Subspecialty','Original Sort Order']

processing_df = response_df[cols_needed].copy()

for cat in survey_cats:
  processing_df[cat] = processing_df['Survey Question 2'].str.contains(cat, \
                                              case=False, na=False, regex=False).astype(int)
processing_df.head()
```
