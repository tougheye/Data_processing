This script takes in the broadstripe export of doctor surveys and maps the responses based on the specialties

```python
import pandas as pd

# read the csv file exported from Broadstripe
file_path = "/content/Survey Responses 11-11 (3).csv"
response_df = pd.read_csv(file_path)
response_df.shape
```

```python
# see which specialties are leading the chart
response_df['Specialty'].value_counts()
```

```python
# number of specialties
response_df['Specialty'].nunique()      #30
```

```python
# get the analysis survey responses in Survey Question 2 field in a list
survey_cats = ['Compensation','Leaves (parental, medical, sick, etc.)',
               'Protected time','Work/life balance','Staffing', 'Retirement']
```

```python
response_df.columns
```

```python
# Create a copy of the repsonse df to work with
cols_needed = ['Broadstripes ID', 'First Name', 'Last Name',\
               'Employer', 'Department', 'Classification','Survey Question 1',\
              'Survey Question 2','Specialty', 'Subspecialty','Original Sort Order']

processing_df = response_df[cols_needed].copy()

# Create columns for each survey category
for cat in survey_cats:
  processing_df[cat] = processing_df['Survey Question 2'].str.contains(cat, \
                                              case=False, na=False, regex=False).astype(int)
processing_df.head()
```

```python
# capitalize specialties and subspecialties columns for better matching

processing_df[['Specialty', 'Subspecialty']] = processing_df[['Specialty', 'Subspecialty']].apply(lambda x: x.str.upper().str.strip())
processing_df.head()
```

```python
processing_df['Specialty'].sort_values().unique()
```

```python
processing_df['Subspecialty'].sort_values().unique()
```

**Family Medicine**

For this category, people tend to list their specialty as Primary Care Provider and Family Medicine, but sometimes they list it as Internal Medicine or maybe another specialty.

WE FILTERED ALL SPECIALTY FIELDS EQUALS "FAMILY MEDICINE" AND SUB-SPECIALTY FIELD containing 'Primary Care'

```python
family_medicine_df = processing_df[(processing_df['Specialty'] == 'FAMILY MEDICINE') |
                processing_df['Subspecialty'].str.contains('PRIMARY CARE', case=False, na=False, regex=False)].copy()
#
```

```python
family_medicine_df.head(160)
```

```python
# prepare the dataframe for excel charts

family_primary_med_sum = family_medicine_df.groupby('Specialty')[survey_cats].sum()
family_primary_med_sum.loc['Total'] = family_primary_med_sum.sum()
```

export dataframes

```python
with pd.ExcelWriter('family_primary_med_chart_prep.xlsx') as writer:
  family_primary_med_sum.to_excel(writer, sheet_name='Family Primary Med summary')
  family_medicine_df.to_excel(writer, sheet_name='Family Primary Med details')
```

```python
specialty_response_cnts = processing_df.groupby('Specialty')[survey_cats].sum()
```

**Hospitalist and Emergecy Medicine**

all specialties with Hospitalist and emergency medicine and subspecialties that contain those terms

```python
Hospitalist_Emergecy_df = processing_df[(processing_df['Specialty'] == 'EMERGENCY MEDICINE') |
                processing_df['Subspecialty'].str.contains('HOSPIT', case=False, na=False, regex=False) |
              processing_df['Subspecialty'].str.contains('EMERGENCY', case=False, na=False, regex=False)].copy()
```

```python
Hospital_Emergecy_summary_df = Hospitalist_Emergecy_df.groupby('Specialty')[survey_cats].sum()
Hospital_Emergecy_summary_df.loc['Total'] = Hospital_Emergecy_summary_df.sum()
```

```python
with pd.ExcelWriter('Hospitalist_Emergecy_chart_prep.xlsx') as writer:
  Hospital_Emergecy_summary_df.to_excel(writer, sheet_name='Hospitalist Emergecy summary')
  Hospitalist_Emergecy_df.to_excel(writer, sheet_name='Hospitalist Emergecy details')
```

**Pathology (physicians and faculty)**

```python

pathology_df = processing_df[(processing_df['Specialty'] == 'PATHOLOGY') |
                processing_df['Subspecialty'].str.contains('PATH', case=False, na=False, regex=False)].copy()
```

```python
pathology_summary_df = pathology_df.groupby('Specialty')[survey_cats].sum()
pathology_summary_df.loc['Total'] = pathology_summary_df.sum()
pathology_summary_df
```

```python
with pd.ExcelWriter('pathology_chart_prep.xlsx') as writer:
  pathology_summary_df.to_excel(writer, sheet_name='Pathology summary')
  pathology_df.to_excel(writer, sheet_name='Pathology details')
```

**Neurology**

```python
neurology_df = processing_df[(processing_df['Specialty'] == 'NEUROLOGY') |
                processing_df['Subspecialty'].str.contains('NEURO', case=False, na=False, regex=False)].copy()
```

```python
neurology_summary_df = neurology_df.groupby('Specialty')[survey_cats].sum()
neurology_summary_df.loc['Total'] = neurology_summary_df.sum()
neurology_summary_df
```

```python
with pd.ExcelWriter('neurology_chart_prep.xlsx') as writer:
  neurology_summary_df.to_excel(writer, sheet_name='Neurology summary')
  neurology_df.to_excel(writer, sheet_name='Neurology details')
```
