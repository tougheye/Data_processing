This project is to take in the medical cost reports and compare the doctor workload data year over year

```python
import pandas as pd

hosp_name = "UCSF_Parnassus"          #update the hospital name
cost_report_file_path = "/content/UCSF_Doc_workload_Anal.xlsx"    #update file path
cost_report_file = pd.ExcelFile(cost_report_file_path)
file_tabs = cost_report_file.sheet_names
file_tabs

```

Read the Doctor count tab

```python
doc_headcount_df = cost_report_file.parse(file_tabs[0], header=[0,1,2])
doc_headcount_df.head()
```

```python
doc_headcount_df.columns = [f"{year}_{hosp}_{cat}" for year, hosp, cat in doc_headcount_df.columns]

# doc_headcount_df.columns
```

```python
column_dict = {'Unnamed: 0_level_0_Unnamed: 0_level_1_Line No.': "Line No",
               'Unnamed: 1_level_0_Unnamed: 1_level_1_Clinical Specialty': "Clinical Specialty",
                '2025_Hospital Based_(1)\nBoard Certified': '2025_Board Certified',
                '2025_Hospital Based_(2)\nBoard Eligible': '2025_Board Eligible',
                '2024_Hospital Based_(1)\nBoard Certified': '2024_Board Certified',
                '2024_Hospital Based_(2)\nBoard Eligible': '2024_Board Eligible',
                '2023_Hospital Based_(1)\nBoard Certified': '2023_Board Certified',
                '2023_Hospital Based_(2)\nBoard Eligible': '2023_Board Eligible',
                '2022_Hospital Based_(1)\nBoard Certified': '2022_Board Certified',
                '2022_Hospital Based_(2)\nBoard Eligible': '2022_Board Eligible',
                '2021_Hospital Based_(1)\nBoard Certified': '2021_Board Certified',
                '2021_Hospital Based_(2)\nBoard Eligible': '2021_Board Eligible',
                '2020_Hospital Based_(1)\nBoard Certified': '2020_Board Certified',
                '2020_Hospital Based_(2)\nBoard Eligible': '2020_Board Eligible'}
```

```python
doc_headcount_df.rename(columns=column_dict, inplace=True)
doc_headcount_df.fillna(0, inplace=True)
doc_headcount_df.head()
```

```python
# REad the cost report category comparison file to get the specialty matches with service categories

category_details = pd.read_excel("/content/drive/MyDrive/Cost_rept_comp_categories.xlsx")
```

```python
category_details['Cost_rev_row'] = category_details['Cost_rev_row'].fillna(0).astype(int)
```

Read the Services data from the updloaded file

```python
service_unit_df = cost_report_file.parse(file_tabs[2], header=[0,1])
# service_unit_df.head()
```

```python
service_unit_df.columns = [f"{year}_{cat}" for year, cat in service_unit_df.columns]
```

```python
col_dict = {'Unnamed: 0_level_0_Line No.': "Line No",
            'Unnamed: 1_level_0_Revenue Producing Centers': "Revenue Producing Centers",
            '2025_(1)\nUnits of Service\nfrom Pg 17 Col 13': "2025_Units of Service",
            '2024_(1)\nUnits of Service\nfrom Pg 17 Col 13': "2024_Units of Service",
            '2023_(1)\nUnits of Service\nfrom Pg 17 Col 13': "2023_Units of Service",
            '2022_(1)\nUnits of Service\nfrom Pg 17 Col 13': "2022_Units of Service",
            '2021_(1)\nUnits of Service\nfrom Pg 17 Col 13': "2021_Units of Service",
            '2020_(1)\nUnits of Service\nfrom Pg 17 Col 13': "2020_Units of Service"}
```

```python
service_unit_df.rename(columns=col_dict, inplace=True)
service_unit_df.head()
```

```python
# cast the line no to int to match the cost report file
service_unit_df['Line No'] = service_unit_df['Line No'].fillna(0).astype(int)
```

```python
serv_unit_col_list = ['Line No', 'Revenue Producing Centers']
for year in range(2025, 2019, -1):
  col_label = f"{year}_Units of Service"
  serv_unit_col_list.append(col_label)
  service_unit_df[col_label] = service_unit_df[col_label].fillna(0).astype(int)

# serv_unit_col_list
```

```python
# doc_per_serv_unit = pd.DataFrame()
col_list = ['Line No', 'Clinical Specialty']
for year in range(2025, 2019, -1):
  col_label = f"{year}_Total_Hosp_doc"
  doc_headcount_df[col_label] = doc_headcount_df[f"{year}_Board Certified"] + doc_headcount_df[f"{year}_Board Eligible"]
  doc_headcount_df[col_label] = doc_headcount_df[col_label].astype(int)
  col_list.append(col_label)

```

**Create the doctor vs workload comparison dataframe**

```python
doc_v_service_df = category_details.merge(doc_headcount_df[col_list], on= 'Clinical Specialty', how='left').rename(columns={'Line No': 'Line No_doc'})
doc_v_service_df = doc_v_service_df.merge(service_unit_df[serv_unit_col_list], left_on='Cost_rev_row', right_on='Line No', how='inner').drop(columns=['Line No_doc', 'Line No',"Revenue Producing Centers"]).drop_duplicates().reset_index(drop=True)
```

Calculate services per doctors

```python
import numpy as np
for year in range(2025, 2019, -1):
  col_label = f"{year}_Total_serv_per_doc"

  denominator = doc_v_service_df[f"{year}_Total_Hosp_doc"]
  # handle the doctor count 0 preventing an infinity
  doc_v_service_df[col_label] = doc_v_service_df[f"{year}_Units of Service"].div(denominator.replace(0, np.nan)).round(2)

```

Add doctor count comparison from 2020 to 2025

```python
from os import replace
denom = doc_v_service_df['2020_Total_Hosp_doc']
nom = doc_v_service_df['2025_Total_Hosp_doc'] - doc_v_service_df['2020_Total_Hosp_doc']
doc_v_service_df['Doc_diff_2020_2025'] = (
    nom.div(denom.replace(0,np.nan))  * 100
).map('{:.2f}%'.format)
```

Set columns in the right order

```python
col_order = ['Clinical Specialty', 'Cost_rev_row', 'Rev_prod_center',
       'Standard Unit\nof Measure', "Doc_diff_2020_2025", '2025_Total_serv_per_doc',
       '2024_Total_serv_per_doc', '2023_Total_serv_per_doc',
       '2022_Total_serv_per_doc', '2021_Total_serv_per_doc',
       '2020_Total_serv_per_doc' ,'2025_Total_Hosp_doc',
       '2024_Total_Hosp_doc', '2023_Total_Hosp_doc', '2022_Total_Hosp_doc',
       '2021_Total_Hosp_doc', '2020_Total_Hosp_doc', '2025_Units of Service',
       '2024_Units of Service', '2023_Units of Service',
       '2022_Units of Service', '2021_Units of Service',
       '2020_Units of Service']
```

```python
doc_v_service_df = doc_v_service_df[col_order]
```

```python
doc_v_service_df.to_excel(f"{hosp_name}_doc_v_service.xlsx")
```
