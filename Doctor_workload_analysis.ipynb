This project is to take in the medical cost reports and compare the doctor workload data year over year

```python
import pandas as pd

cost_report_file_path = "/content/Reagan_Doc_Workload_Anal.xlsx"
cost_report_file = pd.ExcelFile(cost_report_file_path)
file_tabs = cost_report_file.sheet_names

```

```python
doc_headcount_df = cost_report_file.parse(file_tabs[0], header=[0,1])
doc_headcount_df.head()
```

```python
doc_headcount_df.columns
```

```python
doc_headcount_df.columns = [f"{year}_{cat}" for year, cat in doc_headcount_df.columns]

doc_headcount_df.head()
```
