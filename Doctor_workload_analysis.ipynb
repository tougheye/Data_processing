This project is to take in the medical cost reports and compare the doctor workload data year over year

```python
import pandas as pd

cost_report_file_path = "/content/Reagan_Doc_Workload_Anal.xlsx"
cost_report_file = pd.ExcelFile(cost_report_file_path)
file_tabs = cost_report_file.sheet_names

```

```python
doc_headcount_df = cost_report_file.parse(file_tabs[0], header=[0,1])
doc_headcount_df.head()
```

```python
doc_headcount_df.columns
```

```python
doc_headcount_df.columns = [f"{year}_{cat}" for year, cat in doc_headcount_df.columns]

# doc_headcount_df.columns
```

```python
column_dict = {'Unnamed: 0_level_0_Line No.': "Line No", 'Unnamed: 1_level_0_Clinical Specialty': "Clinical Specialty",
       '2025_(1)\nBoard Certified': '2025_Board Certified', '2025_(2)\nBoard Eligible': '2025_Board Eligible',
       '2024_(1)\nBoard Certified': '2024_Board Certified', '2024_(2)\nBoard Eligible': '2024_Board Eligible',
       '2023_(1)\nBoard Certified': '2023_Board Certified', '2023_(2)\nBoard Eligible': '2023_Board Eligible',
       '2022_(1)\nBoard Certified': '2022_Board Certified', '2022_(2)\nBoard Eligible': '2022_Board Eligible',
       '2021_(1)\nBoard Certified': '2021_Board Certified', '2021_(2)\nBoard Eligible': '2021_Board Eligible',
       '2020_(1)\nBoard Certified': '2020_Board Certified', '2020_(2)\nBoard Eligible': '2020_Board Eligible'}
```

```python
doc_headcount_df.rename(columns=column_dict, inplace=True)
doc_headcount_df
```
