```python
import pandas as pd

excel_file = pd.ExcelFile("/content/UPTE_Newly_Accreted_codes.xlsx")
tab_names = excel_file.sheet_names
```

```python
df = excel_file.parse(tab_names[2], dtype=str)

df.head(50)
```

```python
print(df.columns)
```

```python
grouped_df = df.groupby(['Location/Business Unit', 'Title Group']).agg(
    Jobcodes=('Jobcode', lambda x: ', '.join(x)),
    Job_Title_Abbr_Names=('Job Title Abbr Name', lambda x: ', '.join(x))
).reset_index()

display(grouped_df)
```
